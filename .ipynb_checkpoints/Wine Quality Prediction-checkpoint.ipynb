Loading the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

Data Collection

```python
#Loading dataset as pandas dataframe
wine_dataset = pd.read_csv('winequality.csv')

wine_dataset.describe()
```

```python
wine_dataset.isnull().sum()
```

Data Analysis and Visualization

```python
#Number of values for each quality
sns.catplot(x='quality', data=wine_dataset, kind = 'count')
```

```python
#Volatile acidity vs Quality
plot = plt.figure(figsize = (5,5))
sns.barplot(x='quality', y='volatile acidity', data = wine_dataset)
```

```python
#Citric acid vs Quality
plot = plt.figure(figsize = (5,5))
sns.barplot(x='quality', y='citric acid', data = wine_dataset)
```

```python
#Correlation using heatmap
correlation = wine_dataset.corr()
plt.figure(figsize = (8, 8))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f', annot = True, annot_kws={'size': 8}, cmap = 'Blues')
```

Data Processing

```python
X = wine_dataset.drop(columns ='quality')
print(X)
```

Label Binarization


```python
#Change quality values to good 1 or bad 0
Y = wine_dataset['quality'].apply(lambda y_values:1 if y_values>=7 else 0)
print(Y)
```

Model

```python
#Train Test Split
X_train, X_test, Y_train, Y_test = train_test_split(X , Y, stratify=Y, test_size=0.2, random_state=2)

model = RandomForestClassifier()
model.fit(X_train, Y_train)
```


Evaluation

```python
#Training
X_train_prediction = model.predict(X_train)
training_accuracy_score = accuracy_score(X_train_prediction, Y_train)
print('The training accuracy score is:', training_accuracy_score)
```

```python
#Test
X_test_prediction = model.predict(X_test)
test_accuracy_score = accuracy_score(X_test_prediction, Y_test)
print('The test accuracy score is:', test_accuracy_score)
```

Prediction System

```python
input_data = (7.9,0.6,0.06,1.6,0.069,15.0,59.0,0.9964,3.3,0.46,9.4)
input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)
prediction = model.predict(input_data_reshaped)
print(prediction)

if (prediction == 1):
    print('Wine is good')
else:
    print("Wine is bad")


```
